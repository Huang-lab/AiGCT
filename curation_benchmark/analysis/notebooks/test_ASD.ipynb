```python
from aigct.model import VEQueryCriteria
qry1 = VEQueryCriteria(filter_names=['ASD_case1','ASD_case2','ASD_case3','ASD_case4', 'ASD_control1', 'ASD_control2', 'ASD_control3', 'ASD_control4'])
```

```python
from aigct.container import VEBenchmarkContainer

container = VEBenchmarkContainer(CONFIG)

query_mgr = container.query_mgr
CHD = query_mgr.get_variants_by_task('ASD', qry1)
CHD.to_csv(os.path.join(OUTPUT_DIR, "ASD_pri_EVE.csv"))
```

```python
ASD_MAVEN_EVE = pd.read_csv(os.path.join(SCORES_DIR, "ASD_MAVEN_EVE.csv"))
ASD_pri_EVE = pd.read_csv(os.path.join(SCORES_DIR, "ASD_pri_EVE.csv"))
ASD_EVE = pd.concat([ASD_MAVEN_EVE, ASD_pri_EVE])
```

```python
#MAVEN
metrics = container.analyzer.compute_metrics(
    "ASD", ASD_MAVEN_EVE, 'EVE', variant_query_criteria = qry1, variant_effect_sources= ['MAVEN', 'MAVENAVG', 'ALPHAM', 'REVEL', 'ESM1B'], vep_min_overlap_percent=1 ,variant_vep_retention_percent=100)
```

```python
#PRIMATE
metrics = container.analyzer.compute_metrics(
    "ASD", variant_effect_sources= ['PROVEAN', 'MAVENAVG', 'POLYP2HVAR', 'PRIMAI',  'REVEL', 'DEOGEN2', 'CADD', 'VEST4', 'SIFT', 'BAYESDNAF', 'MUTTASTE', 'LISTS2', 'MCAP', 'FATHMMXF', 'DANN'], vep_min_overlap_percent=80 ,variant_vep_retention_percent=100)
```

```python
#ALL
metrics = container.analyzer.compute_metrics(
    "ASD",  vep_min_overlap_percent=90 ,variant_vep_retention_percent=100)
```

```python
container.reporter.write_summary(metrics)
```
